```python
# Exact match baseline: normalized name + address
import sys; sys.path.append("../src")
import pandas as pd
from business_entity_resolution.data.loader import DataLoader
from business_entity_resolution.data.preprocessing import preprocess_dataframe
from business_entity_resolution.evaluation.metrics import evaluate_predictions

loader = DataLoader("../dataset", "train")
s1, s2, s3, gt = loader.load_all_with_truth()
s1, s2, s3 = map(preprocess_dataframe, (s1, s2, s3))
cand = pd.concat([s2, s3])

key = lambda r: (r["name_norm"], r["address_norm"], r["country_norm"])
cand_keys = {key(r): r["entity_id"] for _, r in cand.iterrows()}

pred = []
for _, r in s1.iterrows():
    k = key(r)
    pred.append({"source1_entity_id": r["entity_id"],
                 "matched_entity_ids": cand_keys.get(k, "")})
pred_df = pd.DataFrame(pred)
print(evaluate_predictions(gt, pred_df))
```
